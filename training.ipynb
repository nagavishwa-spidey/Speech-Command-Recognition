# Speech Command Recognition

This notebook trains a CNN model to recognize simple speech commands using the TensorFlow Speech Commands dataset.
It runs end‑to‑end in Google Colab with GPU support.

```python
import tensorflow as tf
import tensorflow_datasets as tfds
import numpy as np
import librosa
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import json, os
```

```python
# Verify GPU
print('GPU available:', tf.config.list_physical_devices('GPU'))
```

```python
# Load the Speech Commands v0.02 dataset (subset of 10 keywords)
splits = tfds.load('speech_commands', split=['train', 'validation', 'test'],
                                 as_supervised=True, with_info=False)
train_ds, val_ds, test_ds = splits
```

```python
# Parameters
SAMPLE_RATE = 16000
DURATION = 1  # seconds
N_MELS = 40
BATCH_SIZE = 64
EPOCHS = 20
```

```python
def preprocess(audio, label):
    # audio: tf.Tensor of shape (None,) containing PCM int16
    audio = tf.cast(audio, tf.float32) / 32768.0  # normalize
    # Ensure fixed length
    audio = tf.pad(audio, [[0, SAMPLE_RATE - tf.shape(audio)[0]]])[:SAMPLE_RATE]
    # Convert to log‑mel spectrogram using librosa
    spect = tf.numpy_function(func=lambda x: librosa.feature.melspectrogram(y=x, sr=SAMPLE_RATE, n_mels=N_MELS),
                                 inp=[audio],
                                 Tout=tf.float32)
    log_mel = tf.numpy_function(func=lambda s: librosa.power_to_db(s, ref=np.max),
                                 inp=[spect],
                                 Tout=tf.float32)
    log_mel = tf.transpose(log_mel)  # shape (time, n_mels)
    log_mel = tf.expand_dims(log_mel, -1)  # add channel dim
    return log_mel, label
```

```python
# Apply preprocessing
train = train_ds.map(preprocess).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
val = val_ds.map(preprocess).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test = test_ds.map(preprocess).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
```

```python
# Build a simple CNN model
input_shape = (int(SAMPLE_RATE/200)+1, N_MELS, 1)  # time steps depend on hop length
model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape=input_shape),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Conv2D(64, (3,3), activation='relu'),
    tf.keras.layers.BatchNormalization(),
    tf.keras.layers.MaxPooling2D((2,2)),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(10, activation='softmax')  # 10 target words
] )
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.summary()
```

```python
# Callbacks
checkpoint_cb = tf.keras.callbacks.ModelCheckpoint('speech_command_model.keras', save_best_only=True)
earlystop_cb = tf.keras.callbacks.EarlyStopping(patience=3, restore_best_weights=True)
```

```python
history = model.fit(train, validation_data=val, epochs=EPOCHS, callbacks=[checkpoint_cb, earlystop_cb])
```

```python
# Plot training history
%matplotlib inline
plt.figure(figsize=(8,4))
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='val')
plt.title('Accuracy')
plt.xlabel('Epoch')
plt.legend()
plt.show()
plt.figure(figsize=(8,4))
plt.plot(history.history['loss'], label='train')
plt.plot(history.history['val_loss'], label='val')
plt.title('Loss')
plt.xlabel('Epoch')
plt.legend()
plt.show()
```

```python
# Evaluate on test set
test_loss, test_acc = model.evaluate(test)
print('Test accuracy:', test_acc)
```

```python
# Detailed classification report
y_true = []
y_pred = []
for x_batch, y_batch in test:
    preds = model.predict(x_batch)
    y_true.extend(y_batch.numpy())
    y_pred.extend(np.argmax(preds, axis=1))
print(classification_report(y_true, y_pred))
cm = confusion_matrix(y_true, y_pred)
import seaborn as sns
plt.figure(figsize=(8,6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

```python
# Save label mapping (TensorFlow DS provides int labels, we map them to words)
info = tfds.builder('speech_commands').info
label_names = info.features['label'].names
with open('labels.json', 'w') as f:
    json.dump({str(i): name for i, name in enumerate(label_names)}, f)
print('Saved labels.json')
```
